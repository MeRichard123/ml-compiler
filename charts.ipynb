```python
# importing libraries 
import matplotlib.pyplot as plt 
import seaborn 

fig, ax = plt.subplots(figsize=(13, 8), subplot_kw=dict(aspect="equal"))
# declaring data 
data = [677 * 0.7, (677 * 0.15) + 1, (677 * 0.15) + 1] 
keys = ['Training', 'Testing', 'Validation'] 
  
# define Seaborn color palette to use 
palette_color = seaborn.color_palette('pastel') 

def func(pct, allvalues):
    absolute = int(pct / 100. * sum(allvalues))
    return f'{pct:.1f}% \n ({absolute:d})'
  
# plotting data on chart 
ax.pie(
    data, 
    labels=keys, 
    colors=palette_color, 
    autopct=lambda pct: func(pct, data),
    textprops=dict(size=16),
    )

# displaying chart 
plt.savefig('data_split.pdf')
```
